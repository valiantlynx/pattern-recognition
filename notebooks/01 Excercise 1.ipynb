```python
# Statistical overview - select only numeric columns. not date etc.
numeric_df = prices_and_sales_df.select_dtypes(include='number')
desc = numeric_df.describe()

# CV = (std / mean) * 100
cv_percent = (numeric_df.std() / numeric_df.mean()) * 100
variance_percent = (numeric_df.var() / (numeric_df.mean() ** 2)) * 100

# Add a new row named 'CV (%)' to the 'desc' DataFrame using .loc[]
desc.loc['Coefficient of Variation (%)'] = cv_percent
desc.loc['Variance (%)'] = variance_percent

desc
```

on the data where the variance is high. sicne the variation is very high, it could mean that those as feature are stronger indicators than for example average price. we will see in the corelation matrix

```python
# univariate analysis of 'Price' column
from scipy import stats
import numpy as np

fig = px.histogram(prices_and_sales_df, x='AveragePrice', 
                   nbins=prices_and_sales_df['AveragePrice'].nunique()//10,
                   title='Distribution of Price',
                   labels={'AveragePrice': 'Price'},
                   marginal='box',
                   histnorm='probability density')

# Calculate KDE for smooth density curve
data = prices_and_sales_df['AveragePrice'].dropna()
kde = stats.gaussian_kde(data)
x_range = np.linspace(data.min(), data.max(), 200)
y_kde = kde(x_range)

fig.add_trace(go.Scatter(x=x_range, y=y_kde, 
                         mode='lines', 
                         name='Density Curve',
                         line=dict(color='red', width=2)))

fig
```

```python
# univariate analysis of 'total volume' column
from scipy import stats
import numpy as np

# I wanna show the density but the deviation is too big. so i use the count to get the density
data = prices_and_sales_df['TotalVolume'].dropna()
log_data = np.log10(data)

fig = px.histogram(x=log_data, 
                   nbins=50,
                   title='Distribution of Total Volume (Log Scale)',
                   labels={'x': 'Log10(Total Volume)'},
                   marginal='box',
                   histnorm='probability density')

# Calculate KDE
kde = stats.gaussian_kde(log_data)
x_range = np.linspace(log_data.min(), log_data.max(), 200)
y_kde = kde(x_range)

fig.add_trace(go.Scatter(x=x_range, y=y_kde, 
                         mode='lines', 
                         name='Density Curve',
                         line=dict(color='red', width=2)))

fig.update_yaxes(title_text='Density')
fig
```

```python
# Bivariate analysis: Average Price variation across years
fig = px.box(prices_and_sales_df, x='year', y='AveragePrice',
             title='Average Price Distribution by Year',
             labels={'year': 'Year', 'AveragePrice': 'Average Price ($)'},
             color='year',
             points='outliers')
fig.update_xaxes(type='category')
fig

```

```python
# Bivariate analysis: Total Volume variation across years (using log scale for density)
# Create a temporary column with log-transformed volume
prices_and_sales_df_copy = prices_and_sales_df.copy()
prices_and_sales_df_copy['Log_TotalVolume'] = np.log10(prices_and_sales_df_copy['TotalVolume'])

fig2 = px.violin(prices_and_sales_df_copy, x='year', y='Log_TotalVolume',
                 title='Total Volume Distribution by Year (Log Scale Density)',
                 labels={'year': 'Year', 'Log_TotalVolume': 'Log10(Total Volume)'},
                 color='year',
                 box=True,  # Add box plot inside violin
                 points='outliers')  # Show outlier points
fig2.update_xaxes(type='category')
fig2
```

```python
# Bivariate analysis: Every variable against every variable (Scatter Matrix)
key_features = ['AveragePrice', 'TotalVolume', '4046', '4225', '4770', 
                'TotalBags', 'SmallBags', 'LargeBags'] # can be changed to use all numeric columns

fig3 = px.scatter_matrix(prices_and_sales_df, 
                         dimensions=key_features,
                         title='Scatter Matrix: Bivariate Analysis of All Variables',
                         height=1200,
                         width=1200)

fig3.update_traces(diagonal_visible=True, showupperhalf=False)

fig3
```

```python
# Heatmap of Correlation Matrix
correlation_matrix = numeric_df.corr()

fig4 = px.imshow(correlation_matrix,
                labels=dict(x="Features", y="Features", color="Correlation"),
                x=correlation_matrix.columns,
                y=correlation_matrix.columns,
                color_continuous_scale='RdBu_r',  # Red for positive, Blue for negative
                zmin=-1,
                zmax=1,
                title='Correlation Matrix Heatmap',
                text_auto='.2f',  # Show correlation values with 2 decimal places
                aspect='auto')

fig4.update_layout(width=900, height=800)
fig4
```
